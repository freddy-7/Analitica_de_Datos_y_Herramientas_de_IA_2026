```python
#Importamos librerías requeridas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sympy import primerange
```

```python
#Convertimos el dataset en dataframe
df= pd.read_csv("Amsterdam.csv")
df
```

```python
#Verificamos información del DataFrame
df.info()
```

```python
#Convertimos el DataFrame de pandas en un arreglo de numpy
array = df.to_numpy()
array
```

***Indexación de un array 1D a partir de un array 2D***

```python
# Acceder a la primera fila de un arreglo de numpy
fila0 = array[0]
fila0
```

```python
# Acceder a la primera fila de un Dataframe de pandas
fila0 = df.iloc[0]
fila0
```

```python
# Acceder a la primera fila de un arreglo de numpy
columna0 = array[:, 0]
columna0
```

***Indexación de múltiples arrays 1D a partir de un array 2D***

```python
#Filas consecutivas 3-5 y la columna 4
Filas_3_a_5= array[3:6, 4]       
Filas_3_a_5
```

```python
#Filas no consecutivas 2, 8, 12 y Todas las columnas
Filas_3_a_5= array[[2, 8, 12], :]       
Filas_3_a_5
```

```python
#Filas combinadas 0, 4-6 y 10 y columna 4
Filas_0_4_5_6_10= array[[0] + list(range(4, 7)) + [10], 4]       
Filas_0_4_5_6_10
```

```python
#Filas pares y todas las columnas
Filas_pares = array[::2, :]
Filas_pares
```

```python
#Filas impares y todas las columnas
Filas_impares = array[1::2, :]
Filas_impares
```

```python
#Filas primas y todas las columnas
primos = list(primerange(0, 20))
Filas_primas = array[primos, :]
Filas_primas
```

```python
#Convertir un arreglo de numpy a CSV
np.savetxt("datos_numpy.csv", array, delimiter=",", fmt="%s", encoding="utf-8")
```

```python
#Convertir un dataframe de pandas a CSV
df.to_csv("datos_pandas.csv")
```
